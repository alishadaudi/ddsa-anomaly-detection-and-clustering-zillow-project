```python
# Code snippet 2
# Importing Anomaly detection ML modules from pycaret
from pycaret.anomaly import *
```

```python
# Code snippet 3
# Importing the zillow/real estate dataset
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/nvamsimohan/DallasDSA/refs/heads/main/Zillow%20Tarrant%20County%20Raw%20Data.csv')

data
```

```python
# Code snippet 4
# Display the data columns
data.columns
```

```python
# Code snippet 5
# Configuring/setting up the anomaly detection algos
anomaly_setup = setup(data)
```

```python
# Code snippet 6
# Building an anomaly detection model using KNN algorithm
knn_model = create_model('knn')
```

```python
# Code snippet 7
# Identifying/predicting the anomalies using the knn model
knn_anomalies = predict_model(knn_model, data)

# Displaying KNN anomalies
knn_anomalies
```

```python
# Code snippet 8
# Displaying ONLY the anomalies
knn_anomalies[knn_anomalies['Anomaly'] == 1]
```

```python
# Code snippet 9
# Exporting all anomalies data to a CSV file
knn_anomalies.to_csv('knn_anomalies.csv')
```

```python
# Code snippet 10
# Importing Clustering ML modules from pycaret
from pycaret.clustering import *
```

```python
# Code snippet 11
# Configuring/setting up the clustering algos
cluster_setup = setup(knn_anomalies)
```

```python
# Code snippet 12
# Building a clustering model using k-Means algorithm
kmeans_model = create_model('kmeans')
```

```python
# Code snippet 13
# Identifying/predicting/assigning the clusters using the kmeans model
kmeans_clusters = assign_model(kmeans_model)

# Displaying Kmeans clusters
kmeans_clusters
```

```python
# Code snippet 14
# Exporting the dataset with anomalies and clusters info
kmeans_clusters.to_csv('Zillow data with Anomalies and Clusters.csv')
```
